# Separable Space-Time: Clock Drift Calculator

Tests Theory #1 from *Beyond Space and Time*: space and time evolve separately, with local clock rate coupled to dark-matter density.

$$\frac{\Delta t}{t} \approx k \frac{\Delta\rho_{DM}}{\rho_0}$$

We don't know k yet — this notebook lets the community fit it from real clock data.

```python
import numpy as np
import matplotlib.pyplot as plt

# --- USER INPUTS ---
k = 1e-3  # coupling constant (to be fit)
rho0 = 0.3  # reference DM density at Earth, GeV/cm^3

# Example locations (illustrative values)
locations = {
    'Earth': 0.300,
    'ISS (400km)': 0.300,
    'GPS orbit': 0.2995,
    'Lunar orbit': 0.299,
    'Mars orbit': 0.298,
    'Deep space (1 AU out of plane)': 0.295,
}

print(f"Testing k = {k:.1e}")
print("\nPredicted fractional drift vs Earth:")
for name, rho in locations.items():
    delta = k * (rho - rho0) / rho0
    print(f"{name:25s} Δt/t = {delta:+.2e}")
```

## Fit k from real data

If you have two clocks with measured drift, solve for k:

```python
def fit_k(measured_drift, rho1, rho2, rho0=0.3):
    """measured_drift = (t2-t1)/t, rho in same units"""
    delta_rho = (rho2 - rho1) / rho0
    return measured_drift / delta_rho

# Example: GPS clocks run fast by ~4.5e-10 vs ground (GR only)
# If we measure an *extra* 1e-15 after GR correction:
extra_drift = 1e-15
k_fit = fit_k(extra_drift, locations['Earth'], locations['GPS orbit'])
print(f"Implied k = {k_fit:.2e}")
```

## Plot predicted drift vs k

```python
ks = np.logspace(-6, -2, 100)
mars_drift = ks * (locations['Mars orbit'] - rho0) / rho0

plt.figure(figsize=(5,3.5))
plt.loglog(ks, np.abs(mars_drift))
plt.axhline(1e-15, color='r', ls='--', label='current clock sensitivity')
plt.xlabel('coupling k')
plt.ylabel('|Δt/t| Earth-Mars')
plt.title('Separable Time Prediction')
plt.grid(True, which='both', ls=':')
plt.legend()
plt.tight_layout()
plt.savefig('k_sensitivity.png', dpi=150)
plt.show()
```
