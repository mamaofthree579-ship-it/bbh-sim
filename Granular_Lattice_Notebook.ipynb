# POST 2 — Granular Spacetime Lattice

Implements $\mathcal{S}=\{x_i\}$ with spacing $a$. Evolve $\psi_i$ on 2D grid.

```python
import numpy as np, matplotlib.pyplot as plt

L=64; a_vals=[0.8,1.5,3.0]; J=1.0; dt=0.02; steps=300

def evolve(a):
    psi = np.random.normal(0,1,(L,L)) + 1j*np.random.normal(0,1,(L,L))
    psi /= np.linalg.norm(psi)
    for _ in range(steps):
        lap = (np.roll(psi,1,0)+np.roll(psi,-1,0)+np.roll(psi,1,1)+np.roll(psi,-1,1)-4*psi)
        # Gaussian cutoff mimics finite a
        psi = psi + 1j*J*np.exp(-0.5/a**2)*lap*dt
    return np.abs(psi)**2

fig,axs = plt.subplots(1,3,figsize=(9,3))
for ax,a in zip(axs,a_vals):
    prob = evolve(a)
    ax.imshow(prob, cmap='inferno'); ax.set_title(f'a={a}'); ax.axis('off')
plt.tight_layout(); plt.savefig('lattice_a.png',dpi=150); plt.show()
```

## Measure coherence length

```python
from scipy.signal import correlate2d

def coherence(prob):
    c = correlate2d(prob-prob.mean(), prob-prob.mean(), mode='same')
    return c[L//2, L//2:].argmax()

for a in a_vals:
    prob = evolve(a)
    print(f'a={a}, coherence ~ {coherence(prob)} pixels')
```
