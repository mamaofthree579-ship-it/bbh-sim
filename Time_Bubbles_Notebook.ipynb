# Theory #3 — Time Bubbles: Simulate Lensing Time Delay

Model: $\Delta t_{lens} = (\beta-1)L/c + \Delta$

```python
import numpy as np, matplotlib.pyplot as plt

c = 3e8 # m/s
# Simulate 20 lens systems with different path lengths
np.random.seed(1)
L = np.random.uniform(1e25,5e25,20) # ~1-5 Gpc in meters
beta_true = 1.000002 # 2 ppm bubble effect
Delta_true = 10*86400 # 10 days offset

t_std = L/c
t_obs = beta_true*t_std + Delta_true
t_obs += np.random.normal(0,0.5*86400,size=len(L)) # 0.5 day noise

plt.scatter(t_std/86400/365, (t_obs-t_std)/86400)
plt.xlabel('Standard delay (yr)'); plt.ylabel('Excess delay (days)'); plt.title('Simulated Bubble Signature'); plt.show()
```

## Fit beta

```python
coeffs = np.polyfit(t_std, t_obs, 1)
beta_fit, Delta_fit = coeffs[0], coeffs[1]
print(f"Fitted beta = {beta_fit:.8f}")
print(f"Delta = {Delta_fit/86400:.2f} days")
print("If beta=1 within errors, no bubble detected.")
```

```python
plt.figure(figsize=(5,3.5))
plt.plot(t_std, t_obs, 'o', label='data')
plt.plot(t_std, beta_fit*t_std+Delta_fit, 'r-', label=f'fit beta={beta_fit:.6f}')
plt.legend(); plt.xlabel('t_std (s)'); plt.ylabel('t_obs (s)'); plt.tight_layout()
plt.savefig('bubble_fit.png',dpi=150); plt.show()
```
