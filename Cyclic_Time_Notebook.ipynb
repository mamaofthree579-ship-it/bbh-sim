# Theory #2 — Cyclic Time: Fit Sinusoids to Clock Data

Model: $t_{obs} = \tau + \sum (A_n/\omega_n)[1-\cos(\omega_n \tau + \phi_n)]$

We look for periodic residuals in timing data.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# --- Generate synthetic clock residuals (replace with real data) ---
np.random.seed(0)
years = np.linspace(0, 10, 500)  # 10 years of data
true_A = 2e-15  # fractional amplitude
true_omega = 2*np.pi/1.0  # 1-year cycle
true_phi = 0.5

residual = (true_A/true_omega) * np.cos(true_omega*years + true_phi)
noise = np.random.normal(0, 5e-16, size=years.shape)
data = residual + noise

plt.plot(years, data*1e15)
plt.xlabel('Years'); plt.ylabel('Residual (fs)'); plt.title('Synthetic Clock Data'); plt.show()
```

## Fit single sinusoid

```python
def cyclic_model(t, A, omega, phi, offset):
    return (A/omega)*np.cos(omega*t + phi) + offset

p0 = [1e-15, 2*np.pi, 0, 0]
popt, pcov = curve_fit(cyclic_model, years, data, p0=p0)
A_fit, omega_fit, phi_fit, off = popt
perr = np.sqrt(np.diag(pcov))

print(f"Fitted A = {A_fit:.2e} ± {perr[0]:.1e}")
print(f"Period = {2*np.pi/omega_fit:.3f} yr")
print(f"If A consistent with 0, no cycle detected.")
```

```python
plt.figure(figsize=(6,3.5))
plt.plot(years, data*1e15, '.', ms=3, label='data')
plt.plot(years, cyclic_model(years, *popt)*1e15, 'r-', label='fit')
plt.legend(); plt.xlabel('Years'); plt.ylabel('Residual (fs)'); plt.tight_layout()
plt.savefig('cyclic_fit.png', dpi=150); plt.show()
```
